## 1. 基础文件操作

```python
import os

# 示例文件路径
demo_file = "demo_test.txt"

# 1. 写入文件
print("1. 写入文件:")
with open(demo_file, 'w', encoding='utf-8') as f:
    f.write("你好，Python!\n")
    f.write("这是第二行内容\n")
    f.write("这是第三行内容\n")
print(f"   ✅ 已写入文件：{demo_file}")
```

```python
# 2. 读取整个文件
print("2. 读取整个文件:")
with open(demo_file, 'r', encoding='utf-8') as f:
    content = f.read()
print(content)
```

```python
# 3. 逐行读取
print("3. 逐行读取:")
with open(demo_file, 'r', encoding='utf-8') as f:
    for i, line in enumerate(f, 1):
        print(f"   第{i}行：{line.strip()}")
```

```python
# 4. 读取为列表
print("4. 读取为列表:")
with open(demo_file, 'r', encoding='utf-8') as f:
    lines = f.readlines()
print(f"   列表内容：{lines}")

# 清理测试文件
os.remove(demo_file)
print(f"\n   ✅ 已清理测试文件：{demo_file}")
```

## 2. 追加文件内容

```python
demo_file = "demo_append.txt"

# 首次写入
with open(demo_file, 'w', encoding='utf-8') as f:
    f.write("第一行内容\n")
print("首次写入完成")

# 追加内容
with open(demo_file, 'a', encoding='utf-8') as f:
    f.write("追加的第二行\n")
    f.write("追加的第三行\n")
print("追加内容完成")

# 读取查看
with open(demo_file, 'r', encoding='utf-8') as f:
    content = f.read()
print(f"最终内容:\n{content}")

# 清理
os.remove(demo_file)
print("✅ 已清理测试文件")
```

## 3. JSON 基础操作

```python
import json

# Python 数据结构
data = {
    "name": "张三",
    "age": 25,
    "city": "北京",
    "skills": ["Python", "Java", "Go"],
    "active": True,
    "salary": None
}

print("1. Python 字典:")
print(f"   原始数据：{data}")
```

```python
# 转换为 JSON 字符串
print("\n2. 序列化为 JSON 字符串:")
json_str = json.dumps(data, ensure_ascii=False, indent=2)
print(f"   JSON 格式:\n{json_str}")
```

```python
# JSON 字符串转回 Python
print("\n3. 反序列化为 Python 对象:")
parsed_data = json.loads(json_str)
print(f"   解析后：{parsed_data}")
print(f"   name 字段：{parsed_data['name']}")
print(f"   skills 列表：{parsed_data['skills']}")
```

## 4. JSON 文件操作

```python
json_file = "demo_data.json"

# 要保存的数据
config = {
    "app_name": "我的应用",
    "version": "1.0.0",
    "debug": True,
    "settings": {
        "theme": "dark",
        "language": "zh-CN",
        "timeout": 30
    }
}

# 写入 JSON 文件
print("1. 写入 JSON 文件:")
with open(json_file, 'w', encoding='utf-8') as f:
    json.dump(config, f, ensure_ascii=False, indent=2)
print(f"   ✅ 已写入：{json_file}")
```

```python
# 读取 JSON 文件
print("\n2. 读取 JSON 文件:")
with open(json_file, 'r', encoding='utf-8') as f:
    loaded_config = json.load(f)

print(f"   应用名称：{loaded_config['app_name']}")
print(f"   版本：{loaded_config['version']}")
print(f"   主题：{loaded_config['settings']['theme']}")
```

```python
# 修改并保存
print("\n3. 修改并保存:")
loaded_config["version"] = "1.1.0"
loaded_config["settings"]["timeout"] = 60

with open(json_file, 'w', encoding='utf-8') as f:
    json.dump(loaded_config, f, ensure_ascii=False, indent=2)
print("   ✅ 已更新配置")

# 清理
os.remove(json_file)
print(f"\n   ✅ 已清理测试文件：{json_file}")
```

## 5. 配置文件读取器类

```python
class ConfigReader:
    """配置文件读取器类"""

    def __init__(self, config_file="config.json"):
        """初始化配置读取器"""
        self.config_file = config_file
        self.config = {}
        self.load()

    def load(self):
        """加载配置文件"""
        try:
            if not os.path.exists(self.config_file):
                print(f"⚠️  配置文件不存在：{self.config_file}")
                self.config = {}
                return False

            with open(self.config_file, 'r', encoding='utf-8') as f:
                self.config = json.load(f)
            print(f"✅ 成功加载配置文件：{self.config_file}")
            return True

        except json.JSONDecodeError as e:
            print(f"❌ JSON 解析错误：{e}")
            self.config = {}
            return False

    def get(self, key, default=None):
        """
        获取配置值（支持点号分隔，如 'database.host'）
        """
        keys = key.split('.')
        value = self.config

        for k in keys:
            if isinstance(value, dict) and k in value:
                value = value[k]
            else:
                return default

        return value

    def set(self, key, value):
        """设置配置值"""
        keys = key.split('.')
        config = self.config

        for k in keys[:-1]:
            if k not in config:
                config[k] = {}
            config = config[k]

        config[keys[-1]] = value

    def save(self):
        """保存配置到文件"""
        try:
            with open(self.config_file, 'w', encoding='utf-8') as f:
                json.dump(self.config, f, ensure_ascii=False, indent=2)
            print(f"✅ 配置已保存：{self.config_file}")
            return True
        except Exception as e:
            print(f"❌ 保存失败：{e}")
            return False

    def show(self):
        """显示所有配置"""
        print(json.dumps(self.config, ensure_ascii=False, indent=2))
```

```python
# 创建示例配置文件
config_file = "demo_config.json"

sample_config = {
    "app": {
        "name": "我的应用",
        "version": "1.0.0",
        "debug": True
    },
    "database": {
        "host": "localhost",
        "port": 5432,
        "name": "mydb"
    },
    "server": {
        "host": "0.0.0.0",
        "port": 8080
    }
}

# 写入示例配置
with open(config_file, 'w', encoding='utf-8') as f:
    json.dump(sample_config, f, ensure_ascii=False, indent=2)
print(f"✅ 创建示例配置文件：{config_file}")
```

```python
# 使用配置读取器
config = ConfigReader(config_file)

print("\n获取配置值:")
print(f"   app.name = {config.get('app.name', '未知')}")
print(f"   app.version = {config.get('app.version', '未知')}")
print(f"   database.host = {config.get('database.host', 'localhost')}")
print(f"   database.port = {config.get('database.port', 3306)}")
print(f"   server.port = {config.get('server.port', 80)}")
```

```python
# 修改配置
print("\n修改配置:")
config.set('app.version', '2.0.0')
config.set('database.password', 'secret123')
config.show()

# 保存配置
config.save()

# 清理
os.remove(config_file)
print(f"\n✅ 已清理测试文件：{config_file}")
```

## 6. 实用工具函数

```python
def read_text_file(filename):
    """读取文本文件"""
    try:
        with open(filename, 'r', encoding='utf-8') as f:
            return f.read()
    except Exception as e:
        print(f"读取文件失败：{e}")
        return None


def write_text_file(filename, content):
    """写入文本文件"""
    try:
        with open(filename, 'w', encoding='utf-8') as f:
            f.write(content)
        return True
    except Exception as e:
        print(f"写入文件失败：{e}")
        return False


def read_json_file(filename):
    """读取 JSON 文件"""
    try:
        with open(filename, 'r', encoding='utf-8') as f:
            return json.load(f)
    except Exception as e:
        print(f"读取 JSON 失败：{e}")
        return None


def write_json_file(filename, data):
    """写入 JSON 文件"""
    try:
        with open(filename, 'w', encoding='utf-8') as f:
            json.dump(data, f, ensure_ascii=False, indent=2)
        return True
    except Exception as e:
        print(f"写入 JSON 失败：{e}")
        return False
```

```python
# 测试工具函数
test_file = "test_utils.txt"

# 写入
success = write_text_file(test_file, "Hello, World!\n你好！")
print(f"写入成功：{success}")

# 读取
content = read_text_file(test_file)
print(f"读取内容：{content}")

# 清理
os.remove(test_file)
```
