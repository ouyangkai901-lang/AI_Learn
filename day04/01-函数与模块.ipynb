## 1. 基础函数定义

```python
# 简单的问候函数
def greet(name):
    """简单的问候函数"""
    print(f"你好，{name}！欢迎来到 Python 世界！")

# 调用函数
greet("张三")
greet("李四")
```

```python
# 带返回值的函数
def greet_with_return(name, time_of_day="今天"):
    """带返回值的问候函数"""
    return f"{time_of_day}好，{name}！祝你学习愉快！"

# 使用返回值
message = greet_with_return("王五")
print(message)

print(greet_with_return("赵六", "下午"))
```

## 2. 函数参数类型

```python
# 位置参数
def positional_args(a, b, c):
    """位置参数示例"""
    print(f"a={a}, b={b}, c={c}")
    return a + b + c

# 按位置传递
result = positional_args(1, 2, 3)
print(f"结果：{result}")
```

```python
# 默认参数
def keyword_args(a, b=10, c=20):
    """默认参数示例"""
    print(f"a={a}, b={b}, c={c}")
    return a + b + c

# 不同的调用方式
print("只传 a:", keyword_args(5))
print("传 a 和 b:", keyword_args(5, 15))
print("关键字参数:", keyword_args(5, c=25, b=20))
```

```python
# 可变参数
def flexible_args(*args, **kwargs):
    """可变参数示例"""
    print(f"位置可变参数：{args}")
    print(f"关键字可变参数：{kwargs}")
    return sum(args)

print("结果:", flexible_args(1, 2, 3, 4, 5))
print("---")
flexible_args(name="Alice", age=25, city="北京")
```

## 3. 多返回值

```python
def multiple_returns(x, y):
    """多返回值示例"""
    sum_val = x + y
    diff_val = x - y
    product_val = x * y
    quotient_val = x / y if y != 0 else None
    return sum_val, diff_val, product_val, quotient_val

# 解包返回值
s, d, p, q = multiple_returns(20, 4)
print(f"20 和 4 的：")
print(f"  和 = {s}")
print(f"  差 = {d}")
print(f"  积 = {p}")
print(f"  商 = {q}")
```

## 4. 作用域

```python
global_counter = 0  # 全局变量

def demonstrate_scope():
    """演示作用域"""
    local_var = "我是局部变量"
    global global_counter
    global_counter += 1
    print(f"  {local_var}")
    print(f"  全局计数器：{global_counter}")
    return local_var

print("第一次调用:")
demonstrate_scope()

print("\n第二次调用:")
demonstrate_scope()

print(f"\n最终全局计数器：{global_counter}")
```

## 5. 高阶函数

```python
def apply_operation(numbers, operation):
    """
    高阶函数：对列表应用操作
    
    Args:
        numbers: 数字列表
        operation: 操作函数
    
    Returns:
        操作后的结果列表
    """
    result = []
    for num in numbers:
        result.append(operation(num))
    return result

def square(x):
    """平方函数"""
    return x * x

def cube(x):
    """立方函数"""
    return x * x * x

# 使用高阶函数
numbers = [1, 2, 3, 4, 5]
print(f"原列表：{numbers}")
print(f"平方：{apply_operation(numbers, square)}")
print(f"立方：{apply_operation(numbers, cube)}")
```

## 6. Lambda 表达式

```python
# 简单的 lambda
add = lambda x, y: x + y
print(f"lambda 加法：5 + 3 = {add(5, 3)}")

# 与 sorted 一起使用
students = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 92},
    {"name": "Charlie", "score": 78},
]

print("\n原始顺序:")
for s in students:
    print(f"  {s['name']}: {s['score']}")

# 按成绩排序
sorted_students = sorted(students, key=lambda x: x["score"], reverse=True)
print("\n按成绩排序（从高到低）:")
for s in sorted_students:
    print(f"  {s['name']}: {s['score']}")
```

## 7. 装饰器基础

```python
import time

def timer_decorator(func):
    """计时装饰器"""
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"  ⏱️ {func.__name__} 执行时间：{(end-start)*1000:.2f}ms")
        return result
    return wrapper

@timer_decorator
def slow_function():
    """模拟慢函数"""
    time.sleep(0.1)
    print("  函数执行完成")

print("调用装饰后的函数:")
slow_function()
```

## 9. 练习：封装猜数字游戏

```python
import random

def guess_number_game():
    """
    封装后的猜数字游戏
    包含完整的异常处理和游戏逻辑
    """
    secret_number = random.randint(1, 100)
    max_attempts = 7
    attempts = 0

    print("=" * 50)
    print("猜数字游戏")
    print(f"我已想好一个 1-100 的数字，你有{max_attempts}次机会")
    print("=" * 50)

    while attempts < max_attempts:
        try:
            user_input = input(f"\n第 {attempts + 1} 次猜测：")

            if user_input.lower() == 'q':
                print(f"游戏结束！答案是 {secret_number}")
                return False

            guess = int(user_input)
            attempts += 1

            if guess == secret_number:
                print(f"\n🎉 恭喜！你用了{attempts}次猜对了！")
                return True
            elif guess < secret_number:
                print("📈 太小了")
            else:
                print("📉 太大了")

        except ValueError:
            print("❌ 请输入有效的数字！")
            attempts -= 1  # 无效输入不消耗机会

    print(f"\n游戏结束！答案是 {secret_number}")
    return False

# 运行游戏（取消注释）
# guess_number_game()
```
