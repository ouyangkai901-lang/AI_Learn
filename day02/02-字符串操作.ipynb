## 1. 字符串切片

```python
# 字符串切片语法：[start:end:step]
text = "Hello, World!"

print(f"原字符串：{text}")
print(f"第一个字符：{text[0]}")      # H
print(f"最后一个字符：{text[-1]}")    # !
print(f"前 5 个字符：{text[0:5]}")     # Hello
print(f"从第 7 个开始：{text[7:]}")    # World!
print(f"倒数 6 个字符：{text[-6:]}")   # World!
print(f"每隔 2 个字符：{text[::2]}")   # Hlo ol!
print(f"反转字符串：{text[::-1]}")    # !dlroW ,olleH
```

## 4. 字符串格式化 - f-string（推荐）

```python
"""
常用格式化选项

  ┌───────────┬─────────────────┬──────────────────────────┐
  │   语法    │      说明       │           示例           │
  ├───────────┼─────────────────┼──────────────────────────┤
  │ {var}     │ 直接输出值      │ f"{name}"                │
  ├───────────┼─────────────────┼──────────────────────────┤
  │ {var:<10} │ 左对齐，宽 10   │ f"{name:<10}"            │
  ├───────────┼─────────────────┼──────────────────────────┤
  │ {var:>10} │ 右对齐，宽 10   │ f"{name:>10}"            │
  ├───────────┼─────────────────┼──────────────────────────┤
  │ {var:^10} │ 居中对齐，宽 10 │ f"{name:^10}"            │
  ├───────────┼─────────────────┼──────────────────────────┤
  │ {var:.2f} │ 浮点数保留 2 位 │ f"{price:.2f}"           │
  ├───────────┼─────────────────┼──────────────────────────┤
  │ {var:05d} │ 整数前补 0      │ f"{num:05d}"             │
  ├───────────┼─────────────────┼──────────────────────────┤
  │ {var=}    │ 显示变量名和值  │ f"{name=}" → name='张三' │
  └───────────┴─────────────────┴──────────────────────────┘
"""

  name = "张三"
  age = 25
  price = 19.9

  # 基本用法
  print(f"我叫{name}")                    # 我叫张三

  # 对齐格式化
  print(f"{'姓名':<10}: {name}")          # 姓名        : 张三
  print(f"{'年龄':>10}: {age}")           #       年龄: 25

  # 数字格式化
  print(f"价格：{price:.2f}")             # 价格：19.90
  print(f"编号：{age:05d}")               # 编号：00025

  # 表达式
  print(f"明年{age + 1}岁")                # 明年 26 岁

  # 调试模式（Python 3.8+）
  print(f"{name=}, {age=}")               # name='张三', age=25
```

## 5. 字符串格式化 - format() 方法

```python
# python 3.6之后用fstring即可，没必要用format()方法，无需过多关注
# format() 方法
print("我叫{}，今年{}岁".format(name, age))

# 指定位置
print("{1}今年{0}岁".format(age, name))

# 指定名称
print("你好，{name}，欢迎学习{course}".format(name="Bob", course="Python"))
```

## 6. 常用字符串方法

```python
text = "  Hello, Python World!  "
print(f"原字符串：'{text}'")

# 去除空白
print(f"去除两端空白：'{text.strip()}'")
print(f"去除左空白：'{text.lstrip()}'")
print(f"去除右空白：'{text.rstrip()}'")

# 大小写转换
print(f"转小写：'{text.lower()}'")
print(f"转大写：'{text.upper()}'")
print(f"首字母大写：'{text.capitalize()}'")

# 查找和替换
print(f"查找 'Python' 位置：{text.find('Python')}")
print(f"是否包含 'Python': {'Python' in text}")
print(f"替换：{text.replace('Python', 'Java')}")

# 分割和连接
words = "apple,banana,orange"
print(f"分割：{words.split(',')}")

fruits = ['apple', 'banana', 'orange']
print(f"连接：{','.join(fruits)}")
```

## 7. 实用示例

```python
# 示例 1：提取文件名和扩展名
filename = "document.pdf"
name = filename.rsplit('.', 1)[0]
ext = filename.rsplit('.', 1)[1]
print(f"文件名：{name}, 扩展名：{ext}")

# 示例 2：格式化输出表格
data = [
    ("Alice", 25, "Engineer"),
    ("Bob", 30, "Manager"),
    ("Charlie", 28, "Designer")
]

print(f"{'姓名':<10} {'年龄':<6} {'职位':<10}")
print("-" * 30)
for name, age, job in data:
    print(f"{name:<10} {age:<6} {job:<10}")
```
