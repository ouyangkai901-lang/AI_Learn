## 1. 变量的定义

```python
# 变量赋值 - Python 中不需要声明类型
name = "Alice"
age = 25
height = 1.68
is_student = False

print(name)
print(age)
print(height)
print(is_student)
```

## 2. 查看数据类型

```python
# 使用 type() 函数查看变量的类型
print(type(name))      # <class 'str'>
print(type(age))       # <class 'int'>
print(type(height))    # <class 'float'>
print(type(is_student)) # <class 'bool'>
```

## 3. 整数类型 (int)

```python
# 整数运算
a = 10
b = 3

print(f"加法：{a} + {b} = {a + b}")
print(f"减法：{a} - {b} = {a - b}")
print(f"乘法：{a} * {b} = {a * b}")
print(f"除法：{a} / {b} = {a / b}")        # 结果是 float
print(f"整除：{a} // {b} = {a // b}")      # 只保留整数部分
print(f"取余：{a} % {b} = {a % b}")
print(f"幂运算：{a} ** {b} = {a ** b}")
```

## 4. 浮点数类型 (float)

```python
# 浮点数
pi = 3.1415926
price = 99.99
temperature = -3.5

print(f"圆周率：{pi}")
print(f"价格：{price}")
print(f"温度：{temperature}")

# 浮点数运算
radius = 5
area = pi * radius ** 2
print(f"半径为{radius}的圆面积：{area}")
```

## 5. 字符串类型 (str)

```python
# 字符串的三种定义方式
s1 = '单引号'
s2 = "双引号"
s3 = """三引号，

可以跨行"""

print(s1)
print(s2)
print(s3)
```

## 6. 布尔类型 (bool)

```python
# 布尔值 - True 和 False（注意首字母大写）
is_true = True
is_false = False

print(f"True 的整数形式：{int(is_true)}")   # 1
print(f"False 的整数形式：{int(is_false)}") # 0

if 0:
    print("hello world")
elif 1 :
    print("hello python")

# 布尔运算
print(f"True and False: {is_true and is_false}")
print(f"True or False: {is_true or is_false}")
print(f"not True: {not is_true}")
```

## 7. 类型转换

```python
# 隐式转换 - Python 自动进行
result = 10 + 3.5  # int + float -> float
print(f"10 + 3.5 = {result}, 类型：{type(result)}")

# 显式转换 - 手动进行
num_str = "123"
num_int = int(num_str)
print(f"字符串 '{num_str}' 转换为整数：{num_int}")

float_num = 3.7
int_num = int(float_num)
print(f"浮点数 {float_num} 转换为整数：{int_num}")  # 直接截断，不四舍五入

# 转换为字符串
age = 25
age_str = str(age)
print(f"整数 {age} 转换为字符串：'{age_str}'")
```
