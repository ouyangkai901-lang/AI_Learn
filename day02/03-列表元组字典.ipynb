## 一、列表 (List)

### 1.1 列表的创建

```python
# 创建列表
fruits = ["apple", "banana", "orange"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True]
empty = []

print(f"水果列表：{fruits}")
print(f"数字列表：{numbers}")
print(f"混合列表：{mixed}")
```

### 1.2 列表索引和切片

```python
fruits = ["apple", "banana", "orange", "grape", "mango"]

# 索引访问
print(f"第一个水果：{fruits[0]}")
print(f"最后一个水果：{fruits[-1]}")

# 切片
print(f"前 3 个水果：{fruits[0:3]}")
print(f"从第 3 个开始：{fruits[2:]}")
print(f"每隔一个：{fruits[::2]}")
```

### 1.3 列表修改

```python
fruits = ["apple", "banana", "orange"]

# 修改元素
fruits[1] = "blueberry"
print(f"修改后：{fruits}")

# 添加元素
fruits.append("banana")  # 末尾添加
print(f"append 后：{fruits}")

fruits.insert(1, "cherry")  # 指定位置插入
print(f"insert 后：{fruits}")

# 删除元素
removed = fruits.pop()  # 移除最后一个
print(f"pop 移除：{removed}, 剩余：{fruits}")

fruits.remove("apple")  # 移除指定值
print(f"remove 后：{fruits}")

del fruits[0]  # 删除指定索引
print(f"del 后：{fruits}")
```

### 1.4 列表常用方法

```python
numbers = [3, 1, 4, 1, 5, 9, 2, 6]

print(f"原列表：{numbers}")
print(f"长度：{len(numbers)}")
print(f"最大值：{max(numbers)}")
print(f"最小值：{min(numbers)}")
print(f"求和：{sum(numbers)}")
print(f"1 出现的次数：{numbers.count(1)}")
print(f"9 的索引：{numbers.index(9)}")

# 排序
numbers.sort()
print(f"排序后：{numbers}")

# 反转
numbers.reverse()
print(f"反转后：{numbers}")
```

## 二、元组 (Tuple)

### 2.1 元组的创建

```python
# 元组 - 不可变的列表
colors = ("red", "green", "blue")
point = (10, 20)
single = (42,)  # 单元素元组需要逗号
empty = ()

print(f"颜色元组：{colors}")
print(f"坐标点：{point}")
print(f"单元素元组：{single}")
```

### 2.2 元组与列表的区别

```python
# 元组可以索引和切片
colors = ("red", "green", "blue", "yellow")
print(f"第一个颜色：{colors[0]}")
print(f"前两个颜色：{colors[0:2]}")

# 但元组不能修改
# colors[0] = "purple"  # 这会报错！

"""
 元组解包（Tuple Unpacking）     

  一句话解释：把元组（或任何可迭代对象）里的元素，一次性赋值给多个变量。
"""
# 元组解包
x, y = point
print(f"x={x}, y={y}")

# 交换变量值
a, b = 1, 2
# 实际发生两步：
  # 1. 右边先打包：(1, 2)
  # 2. 左边解包：a=1, b=2
a, b = b, a  # 利用元组解包交换
print(f"交换后：a={a}, b={b}")
```

## 三、字典 (Dictionary)

### 3.1 字典的创建

```python
# 字典 - 键值对
person = {
    "name": "Alice",
    "age": 25,
    "city": "Beijing"
}

print(f"个人信息：{person}")
print(f"姓名：{person['name']}")
print(f"年龄：{person['age']}")
```

### 3.2 字典的增删改查

```python
person = {"name": "Alice", "age": 25}

# 添加/修改
person["job"] = "Engineer"  # 添加新键值
person["age"] = 26          # 修改现有键
print(f"修改后：{person}")

# 使用 get() 方法（推荐，避免 KeyError）
print(f"城市：{person.get('city', 'Unknown')}")

# 删除
job = person.pop("job")
print(f"移除的工作：{job}")
print(f"删除后：{person}")

del person["age"]
print(f"del 后：{person}")
```

### 3.3 字典常用方法

```python
person = {"name": "Alice", "age": 25, "city": "Beijing"}

# 获取所有键
print(f"所有键：{list(person.keys())}")

# 获取所有值
print(f"所有值：{list(person.values())}")

# 获取所有键值对
print(f"所有键值对：{list(person.items())}")

# 遍历字典
for key, value in person.items():
    print(f"{key}: {value}")
```

## 四、综合示例

```python
# 示例：学生成绩管理系统
students = [
    {"name": "Alice", "scores": [85, 90, 78]},
    {"name": "Bob", "scores": [92, 88, 95]},
    {"name": "Charlie", "scores": [70, 75, 80]}
]

# 计算每个学生的平均分
for student in students:
    avg = sum(student["scores"]) / len(student["scores"])
    print(f"{student['name']} 的平均分：{avg:.1f}")

# 添加新学生
students.append({"name": "David", "scores": [88, 92, 85]})
print(f"\n学生总数：{len(students)}")
```
