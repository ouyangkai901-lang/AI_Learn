## 场景 1：员工信息处理

```python
# 假设我们有一个员工数据列表
employees = [
    {"name": "张三", "age": 28, "department": "技术部", "salary": 15000},
    {"name": "李四", "age": 32, "department": "市场部", "salary": 12000},
    {"name": "王五", "age": 25, "department": "技术部", "salary": 13000},
    {"name": "赵六", "age": 35, "department": "人事部", "salary": 10000},
    {"name": "钱七", "age": 29, "department": "技术部", "salary": 16000}
]

# 任务 1：打印所有员工信息
print("=" * 60)
print("员工信息表")
print("=" * 60)
print(f"{'姓名':<8} {'年龄':<6} {'部门':<8} {'薪资':<10}")
print("-" * 60)

for emp in employees:
    print(f"{emp['name']:<8} {emp['age']:<6} {emp['department']:<8} {emp['salary']:<10}")
```

```python
# 任务 2：计算技术部的平均薪资
tech_employees = [emp for emp in employees if emp["department"] == "技术部"]
tech_salaries = [emp["salary"] for emp in tech_employees]
avg_salary = sum(tech_salaries) / len(tech_salaries)

print(f"技术部人数：{len(tech_employees)}")
print(f"技术部平均薪资：{avg_salary:.2f}元")
```

```python
# 任务 3：找出薪资最高的员工
highest_paid = max(employees, key=lambda x: x["salary"])
print(f"薪资最高的员工：{highest_paid['name']}，薪资：{highest_paid['salary']}元")
```

## 场景 2：销售数据分析

```python
# 月度销售数据
sales_data = {
    "一月": [15000, 23000, 18000, 21000],
    "二月": [17000, 19000, 22000, 25000],
    "三月": [20000, 24000, 19000, 23000]
}

# 任务 1：计算每个月的总销售额
print("每月销售总额：")
for month, sales in sales_data.items():
    total = sum(sales)
    print(f"  {month}: {total}元")
```

```python
# 任务 2：计算所有周的总销售额和平均值
all_sales = []
for week_sales in sales_data.values():
    all_sales.extend(week_sales)

total_sales = sum(all_sales)
avg_weekly_sales = total_sales / len(all_sales)

print(f"\n所有周总销售额：{total_sales}元")
print(f"周平均销售额：{avg_weekly_sales:.2f}元")
```

```python
# 任务 3：找出销售额最高的一周
max_sale = max(all_sales)
print(f"\n最高周销售额：{max_sale}元")
```

## 场景 3：学生成绩统计

```python
# 学生成绩数据
students = [
    {"name": "Alice", "chinese": 85, "math": 92, "english": 88},
    {"name": "Bob", "chinese": 78, "math": 95, "english": 82},
    {"name": "Charlie", "chinese": 92, "math": 88, "english": 90},
    {"name": "David", "chinese": 88, "math": 76, "english": 85},
    {"name": "Eve", "chinese": 95, "math": 98, "english": 96}
]

# 任务 1：为每个学生计算总分和平均分
print("学生成绩统计表")
print("=" * 50)
print(f"{'姓名':<10} {'语文':<6} {'数学':<6} {'英语':<6} {'总分':<6} {'平均分':<8}")
print("=" * 50)

for student in students:
    total = student["chinese"] + student["math"] + student["english"]
    avg = total / 3
    student["total"] = total
    student["avg"] = avg
    print(f"{student['name']:<10} {student['chinese']:<6} {student['math']:<6} {student['english']:<6} {total:<6} {avg:<8.1f}")
```

```python
# 任务 2：找出单科最高分
print("\n单科最高分：")

subjects = ["chinese", "math", "english"]
subject_names = ["语文", "数学", "英语"]

for subj, name in zip(subjects, subject_names):
    top_student = max(students, key=lambda x: x[subj])
    print(f"  {name}: {top_student['name']} ({top_student[subj]}分)")
```

```python
# 任务 3：按总分排名
print("\n总分排名：")
ranked = sorted(students, key=lambda x: x["total"], reverse=True)

for i, student in enumerate(ranked, 1):
    print(f"  第{i}名：{student['name']} - 总分{student['total']}")
```
